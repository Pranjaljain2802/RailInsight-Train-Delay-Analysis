# One final validation before cleaning
---

Let's check whether the existing year and month columns exactly match the date-derived values.

```python
(train["year"] != train["departure_date"].dt.year).sum()
```

```python
(train["month"] != train["departure_date"].dt.month).sum()
```

```python
# 1. Date conversion
train['departure_date'] = pd.to_datetime(train['departure_date'])

# 2. Identifier conversion
train['train_number'] = train['train_number'].astype(str)

# 3. Categorical conversion (Memory saving)
cat_cols = ['train_type', 'season', 'zone', 'zone_abbr', 'source_station_category', 
            'destination_station_category', 'traction_type', 'primary_delay_cause']
for col in cat_cols:
    train[col] = train[col].astype('category')
```

## Data Quality & Validation — Conclusion

The RailInsight dataset contains 1,500,000 rows and 45 columns.

Data quality checks performed:

- No missing values were found.
- No completely duplicated rows were found.
- journey_id is unique for every journey.
- Month values are within the valid range of 1–12.
- Departure hour values are within the valid range of 0–23.
- day_of_week values are within the valid range of 0–6.
- departure_date was converted from string to datetime.
- year and month columns were validated against departure_date.
- No mismatches were found between derived and existing date fields.
- train_number was converted to string because it is an identifier.
- Repeated categorical columns were converted to category dtype.
- is_overloaded contains only the value 0 and therefore has no variation.

Overall, the dataset passed the major data quality and consistency checks and is ready for Exploratory Data Analysis (EDA).
